```python
# wap take the radius of 
# var=radius var=pi=3.14
# print the answer using f sting and format

r=20
pi=3.14
area=pi*r*r
print("area of circle = pi*r*r")
print(f"radius of circle = {r} and pi {pi} area of circle is = {area}")
print("radius of circle =",r,"and pi",pi," area of circle is =", area)
print("radius of circle = {} and pi {} area of circle is = {}".format(r,pi,area))
```

```python
# wap take breath and height of a right angle tringle
# calculate the area
# var1=breadth var2=hight
# formula : 0.5*breadth*hight

breadth=20
height=40
area=0.5*breadth*height
print("area of right angle triangle 0.5 * breadth * height ")
print("breadth = ",breadth,"and hight =",height, "area of right angle triangle = ",area)
print("breadth = {} and hight = {} area of right angle triangle ={}".format(breadth,height,area))
print(f"breadth = {breadth} and hight = {height} area of right angle triangle ={area}")
```

```python
# wap take the bill amount and tip amount
# calculate total bill
# var1: bill amount var2: tip amount
# formula

bill_amount = 2500
tip_amount = 100
total_bill= bill_amount + tip_amount 
print("total amount = bill + tip")
print("bill amount is = ",bill_amount,"and tip amount = ",tip_amount,"and total amount = ",total_bill )
print("bill amount is = {} and tip amount = {} and total amount = {}".format(bill_amount,tip_amount,total_bill))
print(f"bill amount is = {bill_amount} and tip amount = {tip_amount} and total amount = {total_bill}")
```
